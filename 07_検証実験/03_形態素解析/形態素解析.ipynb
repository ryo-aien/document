```python
from sudachipy import tokenizer, dictionary

tokenizer_obj = dictionary.Dictionary().create()
mode = tokenizer.Tokenizer.SplitMode.C
text = "推しのVライバーがリアタイで神展開すぎて泣いた。"
for m in tokenizer_obj.tokenize(text, mode):
    print(m.surface(), m.part_of_speech())
```

## GiNZA（+ spaCy）

```python
!pip install -U ginza ja-ginza
```

```python
import spacy

nlp = spacy.load("ja_ginza")
text = "準防火地域では建蔽率と容積率の制限が厳しい。"
doc = nlp(text)
for token in doc:
    print(token.text, token.lemma_, token.pos_, token.dep_, token.head.text)
```

```python
import subprocess
import MeCab

# パス取得
dicdir = subprocess.run(
    ["mecab-config", "--dicdir"], stdout=subprocess.PIPE
).stdout.decode().strip()
neologd_path = dicdir + "/mecab-ipadic-neologd"

# MeCab + NEologd 初期化
tagger = MeCab.Tagger(f"-d {neologd_path}")
text = "準防火地域では建蔽率と容積率の制限が厳しい。"
print(tagger.parse(text))
```
